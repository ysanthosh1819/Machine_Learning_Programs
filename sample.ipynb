```python
import numpy as np
import pandas as pd
import matplotlib as plt
data = pd.DataFrame({'name': ['Alice', 'Bob', 'Charlie'], 'age': [25, 30, 35]})
df = pd.DataFrame(data)
print(df)
```

```python
import matplotlib.pyplot as plt
plt.plot(df['age'], df['name'], color='blue', marker='o')
plt.title('Age vs Name')
plt.xlabel('Age')
plt.ylabel('Name')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt 
df = sns.load_dataset('planets')
print(df)
df.info()
df.isnull().sum()
df.duplicated().sum()
df.drop_duplicates(inplace=True)
df.duplicated().sum()

sns.scatterplot(x='orbital_period', y='mass', data=df, hue='method', palette='Set1')

plt.title('Orbital Period vs Mass')
plt.xlabel('Orbital Period')
plt.ylabel('Mass')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset('titanic')

# Remove duplicates
data.drop_duplicates(inplace=True)

sns.scatterplot(
    x='age',
    y='fare',
    hue='survived',
    data=data,
    palette='Set1'
)

plt.title('Age vs Fare')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.show()
```
