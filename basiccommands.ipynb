```python
# Basic Python Commands

# Print statement
print("Hello, Python!")

# Variables
name = "Aarav"
age = 20
marks = 85

print("Name:", name)
print("Age:", age)
print("Marks:", marks)

# Arithmetic operations
a = 10
b = 5

print("\nArithmetic Operations")
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Modulus:", a % b)

# Conditional statement
if marks >= 40:
    print("\nResult: Pass")
else:
    print("\nResult: Fail")

# List
subjects = ["Python", "Maths", "Science"]

print("\nSubjects:")
print(subjects)

# Loop
print("\nSubjects using loop:")

for subject in subjects:
    print(subject)
```

```python
# NumPy Array Operations

import numpy as np

# Create NumPy array
marks = np.array([75, 82, 91, 68, 88])

print("NumPy Array:")
print(marks)

# Arithmetic operations
print("\nAddition:")
print(marks + 5)

print("\nSubtraction:")
print(marks - 5)

print("\nMultiplication:")
print(marks * 2)

print("\nDivision:")
print(marks / 2)

# Statistical operations
print("\nStatistical Operations")

print("Maximum:", np.max(marks))
print("Minimum:", np.min(marks))
print("Mean:", np.mean(marks))
print("Median:", np.median(marks))
print("Standard Deviation:", np.std(marks))

# 2D Array
matrix = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("\n2D Array:")
print(matrix)

print("\nShape:")
print(matrix.shape)

print("\nFirst Row:")
print(matrix[0])

print("\nFirst Column:")
print(matrix[:, 0])
```

```python
# Pandas Series

import pandas as pd

# Create Series
marks = pd.Series(
    [75, 82, 91, 68, 88],
    index=[
        "Aarav",
        "Ananya",
        "Arjun",
        "Diya",
        "Rohan"
    ]
)

print("Student Marks:")
print(marks)

print("\nFirst Student Mark:")
print(marks.iloc[0])

print("\nMaximum Mark:")
print(marks.max())

print("\nMinimum Mark:")
print(marks.min())

print("\nAverage Mark:")
print(marks.mean())
```

```python
# Basic Pandas Data Analysis

import pandas as pd

data = {
    "Student Name": [
        "Aarav",
        "Ananya",
        "Arjun",
        "Diya",
        "Rohan"
    ],

    "Age": [
        20,
        19,
        21,
        20,
        19
    ],

    "Study Hours": [
        5,
        7,
        3,
        6,
        4
    ],

    "Total Marks": [
        75,
        82,
        91,
        68,
        88
    ]
}

df = pd.DataFrame(data)

# Display first five rows
print("First Five Rows:")
print(df.head())

# Shape
print("\nShape:")
print(df.shape)

# Information
print("\nDataset Information:")
df.info()

# Statistical description
print("\nStatistical Description:")
print(df.describe())

# Column names
print("\nColumn Names:")
print(df.columns)

# Average marks
print("\nAverage Marks:")
print(df["Total Marks"].mean())

# Highest marks
print("\nHighest Marks:")
print(df["Total Marks"].max())

# Lowest marks
print("\nLowest Marks:")
print(df["Total Marks"].min())
```

```python
# Bar Chart

import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Student Name": [
        "Aarav",
        "Ananya",
        "Arjun",
        "Diya",
        "Rohan"
    ],

    "Total Marks": [
        75,
        82,
        91,
        68,
        88
    ]
}

df = pd.DataFrame(data)

plt.figure(figsize=(8, 5))

plt.bar(
    df["Student Name"],
    df["Total Marks"]
)

plt.xlabel("Student Name")
plt.ylabel("Total Marks")
plt.title("Student-wise Total Marks")

plt.show()
```

```python
# Line Chart

import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Student Name": [
        "Aarav",
        "Ananya",
        "Arjun",
        "Diya",
        "Rohan"
    ],

    "Total Marks": [
        75,
        82,
        91,
        68,
        88
    ]
}

df = pd.DataFrame(data)

plt.figure(figsize=(8, 5))

plt.plot(
    df["Student Name"],
    df["Total Marks"],
    marker="o"
)

plt.xlabel("Student Name")
plt.ylabel("Total Marks")
plt.title("Student Marks Line Chart")

plt.show()
```

```python
# Scatter Plot

import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Student Name": [
        "Aarav",
        "Ananya",
        "Arjun",
        "Diya",
        "Rohan"
    ],

    "Study Hours": [
        5,
        7,
        3,
        6,
        4
    ],

    "Total Marks": [
        75,
        82,
        91,
        68,
        88
    ]
}

df = pd.DataFrame(data)

plt.figure(figsize=(8, 5))

plt.scatter(
    df["Study Hours"],
    df["Total Marks"]
)

plt.xlabel("Study Hours")
plt.ylabel("Total Marks")
plt.title("Study Hours vs Total Marks")

plt.show()
```
